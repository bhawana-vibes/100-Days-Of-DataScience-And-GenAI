```python
import pandas as pd

# DataFrame 1: Employees
df_emp = pd.DataFrame({
    'emp_id': [1, 2, 3, 4],
    'name': ['Amit', 'Priya', 'Rahul', 'Sneha']
})

# DataFrame 2: Departments
df_dept = pd.DataFrame({
    'emp_id': [3, 4, 5, 6],
    'dept': ['HR', 'IT', 'Finance', 'Marketing']
})
result = pd.merge(df_emp, df_dept, on='emp_id', how='inner')
print(result)
```

```python
result = pd.merge(df_emp, df_dept, on='emp_id', how='left')
print(result)
```

```python
result = pd.merge(df_emp, df_dept, on='emp_id', how='right')
print(result)
```

```python
result = pd.merge(df_emp, df_dept, on='emp_id', how='outer')
print(result)
```

```python
pd.merge(df_emp ,df_dept ,left_on='emp_id', right_on='emp_id', how='inner')
```

```python
pd.merge(df_emp,df_dept, on=['emp_id'], how='inner')
```
