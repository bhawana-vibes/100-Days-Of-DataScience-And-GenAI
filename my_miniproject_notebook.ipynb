```python
import sys
import numpy as np

# Python List vs NumPy Array Memory Check
lista = range(100)
arr12 = np.arange(100)

# List की साइज़ देखना
print("Python List Memory:", sys.getsizeof(87) * len(lista))

# NumPy Array की साइज़ देखना
print("NumPy Array Memory:", arr12.itemsize * len(arr12))
```
